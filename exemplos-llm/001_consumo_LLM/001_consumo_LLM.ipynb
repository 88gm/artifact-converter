```python
from dotenv import load_dotenv

load_dotenv()

import anthropic

client = anthropic.Anthropic()
```

```python
response = client.messages.create(
    
    model="claude-haiku-4-5",
    max_tokens=1024,
    messages=[
        {
            "role": "user", 
            "content": "Faça um poema de 2 frases sobre o Sol."
        }
    ]
)

response
```
